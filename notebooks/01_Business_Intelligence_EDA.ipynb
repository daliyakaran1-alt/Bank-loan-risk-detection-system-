# FinSight - Business Intelligence & Exploratory Data Analysis

This notebook orchestrates the enterprise EDA module.

```python
import pandas as pd
import os
import sys
import warnings
warnings.filterwarnings('ignore')

# Add src to path
sys.path.append(os.path.abspath('..'))

from src.analytics.analytics_manager import AnalyticsManager
from src.analytics.eda_engine import EDAEngine
from src.analytics.kpi_engine import KPIEngine
from src.analytics.report_generator import ReportGenerator
from src.analytics.business_insights import BusinessInsightEngine
```

## 1. Load Cleaned Dataset

```python
DATA_PATH = '../data/processed/application_train_cleaned.csv'
print(f'Loading data from {DATA_PATH}...')
# For notebook performance on local dev, load a sample if desired.
df = pd.read_csv(DATA_PATH)
print(f'Loaded dataset with shape: {df.shape}')
```

## 2. Initialize Engines

```python
TARGET_COL = 'TARGET'

analytics_manager = AnalyticsManager(target_col=TARGET_COL)
eda_engine = EDAEngine(target_col=TARGET_COL)
kpi_engine = KPIEngine(target_col=TARGET_COL)
insight_engine = BusinessInsightEngine(target_col=TARGET_COL)
report_gen = ReportGenerator(output_dir='../reports/')
```

## 3. Dataset Overview & Data Profiling

```python
overview = eda_engine.get_dataset_overview(df)
report_gen.generate_dataset_summary_md(overview, 'dataset_summary.md')
print('Dataset summary saved to reports/dataset_summary.md')
```

## 4. Executive KPI Dashboard

```python
kpis = kpi_engine.calculate_all_kpis(df)
report_gen.generate_executive_report(kpis)
print('Executive KPIs PDF generated.')
```

## 5. Automated Business Insights

```python
insights = insight_engine.generate_all_insights(df)
report_gen.generate_business_insights_report(insights)
print('Business Insights PDF generated.')
```

## 6. Generate EDA Charts & Complete HTML Report

```python
# Run all modules
full_results = analytics_manager.run_analytics(df)

# Add EDA specific sections
full_results.sections['Univariate Analysis'] = eda_engine.generate_univariate_analysis(df, max_cols=5)
full_results.sections['Bivariate Analysis'] = eda_engine.generate_bivariate_analysis(df, max_cols=5)
full_results.sections['Multivariate Analysis'] = eda_engine.generate_multivariate_analysis(df)
full_results.sections['Correlation Analytics'] = eda_engine.generate_correlation_analytics(df)

# Generate HTML report
report_gen.generate_eda_html(full_results.sections)
print('EDA HTML Report generated successfully.')
```
